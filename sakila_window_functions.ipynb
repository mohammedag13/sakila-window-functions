## Question 1: Film length ranking

Show each film’s title, length, and position when films are ordered from longest to shortest.

```python
%%sql
SELECT
    title,
    length,
    RANK () OVER (ORDER BY Length desc) AS position
FROM
    film;
```

## Question 2: Rank films within each rating

Show each film’s title, rating, length, and rank compared with films that have the same rating. Rank the longest film first.

```python
%%sql
select 
    title,
    rating,
    length,
    RANK () OVER (PARTITION BY rating ORDER BY length desc) as position
from
    film;
```

# Question 3: Compare each film with its rating’s average
Show every film’s title, rating, and length, plus the average length of films with the same rating. Keep every film in the result.

```python
%%sql
SELECT 
    title,
    rating,
    length,
    AVG(length) OVER (PARTITION BY rating) AS AVG_LENGTH
FROM 
    film;
```

This query shows each film’s title, rating, and length, along with the average length of films with the same rating.
